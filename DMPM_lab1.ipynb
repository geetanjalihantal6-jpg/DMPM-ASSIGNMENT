count the no of vowel and consonant

```python
string = input("Enter a string: ")

vowels = 0
consonants = 0

for ch in string:
    if ch in "aeiouAEIOU":
        vowels += 1
    elif ch != " ":
        consonants += 1

print("Number of vowels =", vowels)
print("Number of consonants =", consonants)
```

check whether a num is prime

```python
num = int(input("Enter a number: "))

if num <= 1:
    print(f"{num} is not a prime number")
elif num == 2:
    print(f"{num} is a prime number")
else:
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{num} is a prime number")
    else:
        print(f"{num} is not a prime number")
```

find the factorial using function

```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fact(n-1)

num = int(input("Enter an integer: "))

if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    result = fact(num)
    print(f"The factorial of {num} is {result}")
```

```python
list1 = [10, 20, 30, 20, 40, 20, 50]

x = int(input("Enter the element to find: "))

count = 0

for i in list1:
    if i == x:
        count += 1

print("Frequency of", x, "=", count)
```

find the maximum and minimum values in a list

```python
list1 = [10, 25, 5, 40, 15]

max = list1[0]
min = list1[0]

for i in list1:
    if i > max:
        max = i
    if i < min:
        min = i

print("Maximum value =", max)
print("Minimum value =", min)
```
