Usage of matplotlib and seaborn in python

Matplotlib is a general-purpose plotting library in Python. It gives you detailed control over almost every part of a graph.

Seaborn is built on top of Matplotlib and is designed especially for statistical data visualization.

It generally produces more attractive statistical graphs with less code.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import kagglehub
path = kagglehub.dataset_download("pavansubhasht/ibm-hr-analytics-attrition-dataset")
```

```python
import pandas as pd
import os

# The 'path' variable from kagglehub.dataset_download points to a directory.
# We need to find the CSV file within that directory.
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
if csv_files:
    data_file_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(data_file_path)
    print(f"Loaded dataset from: {data_file_path}")
    display(df.head())
else:
    print("No CSV files found in the dataset directory.")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Employee Age')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.countplot(data=df, x='Department', hue='Attrition', palette='viridis')
plt.title('Attrition Count by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df, x='YearsAtCompany', y='MonthlyIncome', hue='Attrition', style='Gender', palette='coolwarm', s=100, alpha=0.7)
plt.title('Monthly Income vs. Years at Company (by Attrition and Gender)')
plt.xlabel('Years at Company')
plt.ylabel('Monthly Income')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
avg_income_by_years = df.groupby('YearsAtCompany')['MonthlyIncome'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=avg_income_by_years, x='YearsAtCompany', y='MonthlyIncome', marker='o', color='green')
plt.title('Average Monthly Income by Years at Company')
plt.xlabel('Years at Company')
plt.ylabel('Average Monthly Income')
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(avg_income_by_years['YearsAtCompany'].unique())
plt.tight_layout()
plt.show()
```

```python
attrition_counts = df['Attrition'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(attrition_counts, labels=attrition_counts.index, autopct='%1.1f%%', startangle=90, colors=['#66b3ff', '#ff9999'])
plt.title('Overall Employee Attrition Distribution')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
import kagglehub
path = kagglehub.dataset_download("rahulsah06/titanic")
```

```python
import os

# The 'path' variable from kagglehub.dataset_download points to a directory.
# Let's list the files in the directory to find the correct CSV filename.
files_in_path = os.listdir(path)
print(f"Files found in the dataset directory: {files_in_path}")

# Common filenames for Titanic dataset are 'train.csv' or 'titanic_train.csv'
# We'll try to load 'train.csv' first.
titanic_csv_filename = 'train.csv'
titanic_data_file_path = os.path.join(path, titanic_csv_filename)

if os.path.exists(titanic_data_file_path):
    df_titanic = pd.read_csv(titanic_data_file_path)
    print(f"Loaded Titanic dataset from: {titanic_data_file_path}")
    display(df_titanic.head())
else:
    print(f"Error: The file '{titanic_csv_filename}' was not found in '{path}'.")
    print("Please check the listed files above and update 'titanic_csv_filename' accordingly.")
    # If 'train.csv' isn't found, you might need to adjust titanic_csv_filename
    # to another name like 'titanic_train.csv' or the one you find in 'files_in_path'.
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df_titanic, x='Survived', y='Age', palette='Set2')
plt.title('Age Distribution by Survival Status on the Titanic')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
def reverse_string(s):
    s = list(s)
    left = 0
    right = len(s)-1

    while left < right:
        s[left],s[right] = s[right],s[left]

        left +=1
        right -=1
    return ''.join(s)

ans=reverse_string("hello")
print(ans)
```

```python
def move_zero(num):
    j = 0
    for i in range (len(num)):
        if num[i] != 0:
            num[j] =num[i]

            j+=1
    while j<len(num):
                num[j] =0

                j+=1
    return num
ans=move_zero([5,0,6,0,2,1])
print(ans)
```
