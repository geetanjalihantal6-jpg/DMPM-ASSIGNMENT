Implement min-max normalization on a set of data points

```python
# Min-Max Normalization in Python

data = [10, 20, 30, 40, 50]

minimum = min(data)
maximum = max(data)

normalized_data = []

for x in data:
    normalized_value = (x - minimum) / (maximum - minimum)
    normalized_data.append(normalized_value)

print("Original data:", data)
print("Normalized data:", normalized_data)
```

Implement the min-max normalization from kaggle datset

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import kagglehub
path = kagglehub.dataset_download("arashnic/fitbit")
```

```python
import os

# The 'path' variable is '/kaggle/input/fitbit', which is a directory.
# The actual CSV files are in subdirectories like 'mturkfitbit_export_3.12.16-4.11.16'

# Let's target the first subdirectory and assume 'Fitbit.csv' is inside it
sub_directory = 'mturkfitbit_export_3.12.16-4.11.16'
full_sub_path = os.path.join(path, sub_directory)

files_in_sub_directory = os.listdir(full_sub_path)
print(f"Files found in the subdirectory '{full_sub_path}': {files_in_sub_directory}")

csv_file_name = None
for file in files_in_sub_directory:
    if file.lower() == 'fitbit.csv': # Check for case-insensitive match
        csv_file_name = file
        break

if csv_file_name:
    full_csv_path = os.path.join(full_sub_path, csv_file_name)
    print(f"Attempting to read CSV from: {full_csv_path}")
    df = pd.read_csv(full_csv_path)
    display(df.head())
else:
    print(f"Error: 'Fitbit.csv' (or 'fitbit.csv') not found in the subdirectory '{full_sub_path}'. Found: {files_in_sub_directory}")
```

Let's choose the `Steps` column for normalization. We'll reshape it as `MinMaxScaler` expects a 2D array.

```python
data_to_normalize = df[['Steps']]

scaler = MinMaxScaler()
normalized_data_sklearn = scaler.fit_transform(data_to_normalize)

df['Steps_Normalized_Sklearn'] = normalized_data_sklearn

print("Original 'Steps' data (first 5 rows):")
display(df['Steps'].head())
print("\nNormalized 'Steps' data using sklearn (first 5 rows):")
display(df['Steps_Normalized_Sklearn'].head())
```

We can also implement the min-max normalization manually to verify the results with the `MinMaxScaler`.

```python
min_steps = df['Steps'].min()
max_steps = df['Steps'].max()

df['Steps_Normalized_Manual'] = (df['Steps'] - min_steps) / (max_steps - min_steps)

print("Normalized 'Steps' data manually (first 5 rows):")
display(df['Steps_Normalized_Manual'].head())
```

implement the decimal scaling normalization on a set of data points

```python
import numpy as np

# Using the 'data' list from the previous example
# data = [10, 20, 30, 40, 50]

# Calculate the maximum absolute value in the data
max_abs_value = np.max(np.abs(data))

# Determine 'j' such that 10^j >= max_abs_value
j = 0
while 10**j <= max_abs_value:
    j += 1

# Perform decimal scaling normalization
decimal_scaled_data = [x / (10**j) for x in data]

print("Original data:", data)
print(f"Maximum absolute value: {max_abs_value}")
print(f"Scaling factor (10^j): {10**j}")
print("Decimal scaled data:", decimal_scaled_data)
```

implement the z-score normalization on 10 data points entered by the user

```python
import numpy as np

# You can enter your 10 data points here
user_data_points_str = input("Enter 10 numerical data points separated by commas ")

try:
    # Convert the string input to a list of floats
    user_data_points = [float(x.strip()) for x in user_data_points_str.split(',')]

    if len(user_data_points) != 10:
        print(f"Error: You entered {len(user_data_points)} data points. Please enter exactly 10.")
    else:
        print("User-entered data points:", user_data_points)
except ValueError:
    print("Error: Invalid input. Please enter numerical values separated by commas.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
# Ensure user_data_points is defined and has 10 elements before proceeding
if 'user_data_points' in locals() and len(user_data_points) == 10:
    # Calculate the mean and standard deviation
    mean_data = np.mean(user_data_points)
    std_dev_data = np.std(user_data_points)

    # Apply Z-score normalization
    z_score_normalized_data = [(x - mean_data) / std_dev_data for x in user_data_points]

    print("Original user data points:", user_data_points)
    print(f"Mean: {mean_data:.2f}")
    print(f"Standard Deviation: {std_dev_data:.2f}")
    print("Z-score normalized data:", [f'{x:.4f}' for x in z_score_normalized_data])
else:
    print("Please run the previous cell and enter 10 valid numerical data points.")
```
