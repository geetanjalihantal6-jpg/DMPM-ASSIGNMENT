```python
# WAP to find the mean of data items (without using a function)

n = int(input("Enter number of data items: "))
total = 0

for i in range(n):
    num = int(input("Enter number: "))
    total = total + num

mean = total/n

print("Mean =", mean)
```

```python
# WAP to find the mean of data items using a function

def mean(data):
    return sum(data) / len(data)

n = int(input("Enter number of data items: "))
data = []

for i in range(n):
    data.append(int(input("Enter number: ")))

print("Mean =", mean(data))
```

```python
# WAP to find the median of data items (without using a function)

n = int(input("Enter number of data items: "))
data = []

for i in range(n):
    num = int(input("Enter number: "))
    data.append(num)

data.sort()   # Sort the data

if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]

print("Median =", median)
```

```python
# WAP to find the Median using a Function

def median(data):
    data.sort()
    n = len(data)
    if n % 2 == 0:
        return (data[n//2-1] + data[n//2]) / 2
    else:
        return data[n//2]

n = int(input("Enter number of data items: "))
data = []

for i in range(n):
    data.append(int(input("Enter number: ")))

print("Median =", median(data))
```

```python
# WAP to find the Mode of data items (without using a function)

n = int(input("Enter number of data items: "))
data = []

for i in range(n):
    data.append(int(input("Enter number: ")))

mode = data[0]
max_count = 0

for i in data:
    count = data.count(i)
    if count > max_count:
        max_count = count
        mode = i

print("Mode =", mode)
```

```python
# WAP to find the Mode of data items using a Function

def mode(data):
    m = data[0]
    max_count = 0

    for i in data:
        count = data.count(i)
        if count > max_count:
            max_count = count
            m = i
    return m

n = int(input("Enter number of data items: "))
data = []

for i in range(n):
    data.append(int(input("Enter number: ")))

print("Mode =", mode(data))
```
